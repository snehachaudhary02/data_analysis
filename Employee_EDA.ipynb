```python
# Import Libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Load Data file
xls = pd.ExcelFile("Cleaned_Employee_Dataset_10000.xlsx")
df = pd.read_excel(xls, sheet_name=xls.sheet_names[0])
```

```python
# Fix acronym casing for IT and HR
df["Department_Clean"] = df["Department_Clean"].replace({"It": "IT", "Hr": "HR"})
df["Join_Date_Clean"] = pd.to_datetime(df["Join_Date_Clean"])
```

```python
# Feature engineering
ref_date = pd.Timestamp("2026-11-21")
df["Tenure_Days"] = (ref_date - df["Join_Date_Clean"]).dt.days
df["Tenure_Years"] = (df["Tenure_Days"] / 365.25).round(2)

df["Age_Group"] = pd.cut(
    df["Age_Clean"],
    bins=[17, 29, 39, 65],
    labels=["20s (18-29)", "30s (30-39)", "40+ (40-65)"],
)
df["Salary_Tier"] = pd.cut(
    df["Salary_Clean"],
    bins=[0, 50000, 90000, 1000000],
    labels=["Entry/Core (<=50k)", "Mid (50k-90k)", "Senior (>90k)"],
)
df["Full_Name_Clean"] = (
    df["First_Name_Clean"].astype(str) + " " + df["Last_Name_Clean"].astype(str)
)
```

```python
# Group summaries
dept_summary = (
    df.groupby("Department_Clean", observed=False)
    .agg(
        Employees=("Customer_ID", "count"),
        Avg_Salary=("Salary_Clean", "mean"),
        Median_Salary=("Salary_Clean", "median"),
        Avg_Age=("Age_Clean", "mean"),
        Avg_Rating=("Performance_Rating_Clean", "mean"),
        Avg_Tenure_Yrs=("Tenure_Years", "mean"),
    )
    .sort_values("Employees", ascending=False)
)

city_summary = (
    df.groupby("City_Clean", observed=False)
    .agg(
        Employees=("Customer_ID", "count"),
        Avg_Salary=("Salary_Clean", "mean"),
        Avg_Rating=("Performance_Rating_Clean", "mean"),
    )
    .sort_values("Employees", ascending=False)
)

print(dept_summary.round(2))
print(city_summary.round(2))
print(
    df[
        [
            "Age_Clean",
            "Salary_Clean",
            "Performance_Rating_Clean",
            "Tenure_Years",
        ]
    ]
    .corr()
    .round(4)
)

df.to_excel("Analyzed_Employee_Dataset_10000.xlsx", index=False)
```
